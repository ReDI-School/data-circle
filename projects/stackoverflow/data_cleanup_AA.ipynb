```python
import sqlite3
import pandas as pd

conn = sqlite3.connect(":memory:")
df.to_sql("new_file", conn, index=False)

columns = [
    "ResponseId", "Age", "EdLevel", "Employment", "WorkExp", "YearsCode",
    "DevType", "OrgSize", "ICorPM", "RemoteWork", "Industry", "AIThreat",
    "NewRole", "ToolCountWork", "ToolCountPersonal", "Country", "Currency",
    "CompTotal", "AISelect", "AIAgents", "ConvertedCompYearly", "JobSat"
]

# Keep only columns that actually exist in df
existing = [c for c in columns if c in df.columns]

query = f"SELECT {', '.join(existing)} FROM new_file"
result = pd.read_sql_query(query, conn)

display(result.head(10))
```

```python
df["Age"] = df["Age"].str.replace("years old", "").str.strip()
df["Age"] = df["Age"].str.replace("years or older", "- over").str.strip()
```

```python
df.groupby("EdLevel")["ResponseId"].count().sort_values(ascending=False)
```

```python
df["EdLevel"] = df["EdLevel"].str.replace(r"^Bachelor.*$", "Bachelor", regex=True)
df["EdLevel"] = df["EdLevel"].str.replace(r"^Master.*$", "Master", regex=True)
df["EdLevel"] = df["EdLevel"].str.replace(r"Secondary school.*$", "Secondary school", regex=True)
df["EdLevel"] = df["EdLevel"].str.replace(r"Some college.*$", "Some college/university", regex=True)
df["EdLevel"] = df["EdLevel"].str.replace(r"Professional.*$", "PhD", regex=True)
df["EdLevel"] = df["EdLevel"].str.replace(r"Associate.*$", "Associate", regex=True)
df["EdLevel"] = df["EdLevel"].str.replace(r"Other.*$", "Other", regex=True)

```

###### For Clarification purposes the column "EdLevel" implying "Education Level" has been totally replaced with some other definitions they are as follows:

###### 1. Bachelor’s degree (B.A., B.S., B.Eng., etc.) = Bachelor
###### 2. Master’s degree (M.A., M.S., M.Eng., MBA, etc.)  = Master
###### 3. Some college/university study without earning a degree = Some college/university
###### 4. Secondary school (e.g. American high school, German Realschule or Gymnasium, etc.) = Secondary school
###### 5. Professional degree (JD, MD, Ph.D, Ed.D, etc.) = PhD
###### 6. Associate degree (A.A., A.S., etc.) = Associate
###### 7. Other (please specify) = Other
###### 8. Primary/elementary school = Primary/elemntary school
